## Task 07: String Validation Methods

**(a)** startswith(), endswith(), and find().

```python
text = "NLP is the future of AI research"

if text.startswith("NLP"):
    print("The text begins with 'NLP'.")
else:
    print("The text does not begin with 'NLP'.")

if text.endswith("research"):
    print("The text ends with 'research'.")
else:
    print("The text does not end with 'research'.")

print(text.find("future"))
```

**(b)** Classify tokens as alphabetic, numeric, or mixed.

```python
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for token in tokens:
    if token.isalpha():
        print(f"{token}: purely alphabetic")
    elif token.isdigit():
        print(f"{token}: purely numeric")
    else:
        print(f"{token}: mixed")
```

**(c)** Separate words into pure_words, pure_numbers, and mixed_tokens.

```python
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for w in words:
    if w.isalpha():
        pure_words.append(w)
    elif w.isdigit():
        pure_numbers.append(w)
    else:
        mixed_tokens.append(w)

print(pure_words)
print(pure_numbers)
print(mixed_tokens)
```

**(d)** Write validate_token() and apply it to a list of tokens.

```python
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

test_tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]
for t in test_tokens:
    print(t, "->", validate_token(t))
```

## Task 08: Functions and Lambda for NLP Pipelines

**(a)** Regular function clean() and an equivalent lambda clean_lambda.

```python
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

sample = "  HELLO Natural Language Processing  "
print(clean(sample))
print(clean_lambda(sample))
```

**(b)** tokenize() and remove_short(), chained together.

```python
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

sentence = "I love working with NLP tools in Python daily"
result = remove_short(tokenize(sentence), 3)
print(result)
```

**(c)** word_lengths() returns a dictionary of word -> length.

```python
def word_lengths(tokens):
    return {t: len(t) for t in tokens}

tokens = ["language", "model", "ai", "processing", "deep"]
print(word_lengths(tokens))
```

**(d)** Combine into a full nlp_pipeline() function.

```python
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered = remove_short(tokens, min_len)
    return word_lengths(filtered)

print(nlp_pipeline("Natural Language Processing helps machines understand text", 3))
print(nlp_pipeline("  Deep Learning models require a lot of data  ", 2))
```

## Task 09: map(), filter(), sorted() and Counter

**(a)** Use map() to lowercase words and compute their lengths.

```python
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

lower_words = list(map(lambda w: w.lower(), words))
print(lower_words)

word_lens = list(map(lambda w: len(w), words))
print(word_lens)
```

**(b)** Use filter() to select long words and words containing "t".

```python
words_long = list(filter(lambda w: len(w) > 3, words))
print(words_long)

words_with_t = list(filter(lambda w: "t" in w.lower(), words))
print(words_with_t)
```

**(c)** Sort the words list three different ways.

```python
alphabetical = sorted(words)
print("Alphabetical:", alphabetical)

by_length_asc = sorted(words, key=len)
print("By length (ascending):", by_length_asc)

by_length_desc = sorted(words, key=len, reverse=True)
print("By length (descending):", by_length_desc)
```

**(d)** Use Counter to compute word frequencies.

```python
from collections import Counter

sentence = "nlp nlp python ai nlp python text ai nlp python text"
words_split = sentence.split()
freq = Counter(words_split)

print(freq)
print(freq.most_common(3))
print(freq["python"])
print(len(words_split), len(freq))
```
